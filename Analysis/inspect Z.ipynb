# Inspecting differences

```python
adm_A = set(mimicdata.loc[mimicdata["A"].eq(1), "admission_id"].unique())
adm_Z = set(Zdata["admission_id"].unique())

missing = sorted(adm_A - adm_Z)
extra   = sorted(adm_Z - adm_A)

len(missing), missing[:20], len(extra), extra[:20]
```

```python
len(adm_A)
```

```python
len(adm_Z)
```

```python
mZ = mimicdata.groupby("admission_id")["Z"].max()
zZ = Zdata.groupby("admission_id")["Z"].max()

diff_ids = mZ.index[(mZ == 1) & (zZ == 0)]
len(diff_ids), diff_ids[:10]
```

```python
disc = (mimicdata.loc[mimicdata["A"] == 1, ["admission_id", "t0", "death_time_from_intime"]]
        .sort_values(["admission_id","t0"])
        .drop_duplicates("admission_id", keep="first")
        .set_index("admission_id"))

pd.Series({
    "diff_count": len(diff_ids),
    "death_td_missing_on_discharge_among_diff": disc.loc[diff_ids, "death_time_from_intime"].isna().mean(),
    "death_td_missing_on_discharge_overall": disc["death_time_from_intime"].isna().mean(),
})
```

```python
diff_df = mimicdata.loc[mimicdata["admission_id"].isin(diff_ids),
                        ["admission_id","death_time_from_intime"]].drop_duplicates()

hours = diff_df["death_time_from_intime"] / pd.Timedelta(hours=1)

# check how close they are to multiples of 12
remainder = hours % 12

print("Near 0h remainder (<1e-6):",
      (remainder.abs() < 1e-6).mean())
```
