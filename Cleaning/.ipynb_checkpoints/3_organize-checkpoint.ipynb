## Drop rows whose observation/recorded time "time" is within 5hrs of discharge "outtime"

```python
def drop_near_discharge(df, buffer=LEAK_BUFFER):
    df = df.copy()
    df["time"] = pd.to_timedelta(df["time"])
    df["los"]  = pd.to_timedelta(df["los"])
    return df.loc[df["time"] <= df["los"] - buffer]
ltm_noleak = drop_near_discharge(ltm)
```

```python
near_discharge_after = ltm_noleak[ltm_noleak["time"] > (ltm_noleak["los"] - LEAK_BUFFER)]

print("Rows within last 5h after filtering:", len(near_discharge_after))
```

```python
# Columns that do NOT change over time (baseline variables)
baseline_cols = [
    "stay_id", "subject_id", "raw_age", "sex", "raw_height", "raw_weight",
    "unit_type", "origin", "los", "intime", "outtime", "death_time_from_intime", "icu_mortality"
]

# Pivot: variables → columns, values → entries
# 1. Make the (stay_id, time, variable) key truly unique by averaging duplicates.
dedup = (ltm_noleak
         .groupby(["stay_id","time","variable"], as_index=False, sort=False)
         ["value"].mean())
n_dups = len(ltm_noleak) - len(dedup)
print(f"Collapsed {n_dups} duplicate (stay_id,time,variable) rows by mean.")

# 2. Now pivot — every cell is single-valued.
ltm_wide = dedup.pivot(index=["stay_id","time"],
                       columns="variable", values="value").reset_index()


# Merge baseline columns back (they are constant per stay_id)
ltm_wide = ltm_wide.merge(
    ltm_noleak[baseline_cols].drop_duplicates("stay_id"),
    on="stay_id",
    how="left"
)
```

```python
assert "glasgow_coma_score" not in ltm_wide.columns or \
       ltm_wide["glasgow_coma_score"].isna().all(), \
       "A precomputed glasgow_coma_score column exists and would be overwritten."

# Combine GCS
ltm_wide["glasgow_coma_score"] = (
    ltm_wide["glasgow_coma_score_eye"]
    + ltm_wide["glasgow_coma_score_motor"]
    + ltm_wide["glasgow_coma_score_verbal"]
)

# Replace invalid sums with NaN if any component is missing
ltm_wide["glasgow_coma_score"] = (
    ltm_wide["glasgow_coma_score"]
    .where(
        ltm_wide[[
            "glasgow_coma_score_eye",
            "glasgow_coma_score_motor",
            "glasgow_coma_score_verbal"
        ]].notna().all(axis=1)
    )
)

#ltm_wide['glasgow_coma_score'] = ltm_wide['glasgow_coma_score'].fillna(15)
```

```python
# Combine ABP
df = ltm_wide.copy()
# ---------------------------
# 1. Compute invasive MAP if SBP/DBP are present
# ---------------------------
df["MAP_art_calc"] = (
    df["invasive_systolic_blood_pressure"] +
    2 * df["invasive_diastolic_blood_pressure"]
) / 3

# Pick invasive_mean if exists, else computed invasive
df["MAP_invasive"] = df["invasive_mean_blood_pressure"].combine_first(df["MAP_art_calc"])


# ---------------------------
# 2. Compute non-invasive MAP if SBP/DBP are present
# ---------------------------
df["MAP_nibp_calc"] = (
    df["noninvasive_systolic_blood_pressure"] +
    2 * df["noninvasive_diastolic_blood_pressure"]
) / 3

# Pick noninvasive_mean if exists, else computed non-invasive
df["MAP_noninvasive"] = df["noninvasive_mean_blood_pressure"].combine_first(df["MAP_nibp_calc"])

for col in ["MAP_invasive", "MAP_noninvasive"]:
    df.loc[(df[col] <= 0) | (df[col] > 250), col] = np.nan

# ---------------------------
# 3. Final clinically informed hierarchy
# ---------------------------
# Use invasive MAP when available, else non-invasive MAP
df["arterial_blood_pressure"] = df["MAP_invasive"].combine_first(df["MAP_noninvasive"])


# Optional: cleanup intermediate columns if you want
df.drop(columns=[
    "MAP_art_calc",
    "MAP_invasive",
    "MAP_nibp_calc",
    "MAP_noninvasive"
], inplace=True)

ltm_wide["arterial_blood_pressure"] = df["arterial_blood_pressure"].copy()
```

```python
# O2 saturation
# Clinically correct hierarchy: SaO₂ (ABG) > SpO₂ (pulse ox)
ltm_wide["oxygen_saturation"] = ltm_wide["O2_arterial_saturation"] \
    .combine_first(ltm_wide["O2_pulseoxymetry_saturation"])

# Cut O2 saturation to 100%
ltm_wide["oxygen_saturation"] = (
    ltm_wide["oxygen_saturation"].clip(upper=100)
)
```

## Clinically meaningful clipping and cleaning
#### 1. Physiology possible range
#### 2. ICU-plausible range

### Clinical logic to use:
#### Temperature, SpO₂, FiO₂, HR, RR, MAP: bounded by physiology/monitoring, so clipping can be fairly rule-based.
#### Lactate, creatinine, ureum, aPTT, PaO₂: real ICU extremes can be very high, so upper clipping must be conservative.
#### Urine output: zero can be real; very high values often reflect documentation/binning issues, so this needs special handling.
#### ABG variables (bicarbonate, PaCO₂, PaO₂): use broad ICU-plausible limits, not normal outpatient ranges. Normal adult ABG reference values are roughly HCO₃⁻ 22–26 mEq/L, PaCO₂ 35–45 mmHg, PaO₂ 80–100 mmHg, but clipping should be much wider in ICU data.

Preprocessign rule:
1. Sentinel/Unit errors -> NA
2. Impossible values -> NA
3. Severe but plausible ICU values -> Keep. But, define a clinical boundary. Beyond which declare missing (NA). Techniques like MICE can be applied

```python
# "set_missing": values outside bounds -> NaN
# "clip": values outside bounds -> truncated to boundary

L_to_clip = [
 'bicarbonate',
 'PTT',
 'temperature',
 'hemoglobin',
 'heart_rate',
 'arterial_blood_pressure',
 'creatinine',
 'blood_urea_nitrogen',
 'urine_output',
 'lactate',
 #'glasgow_coma_score', already handled above
 'paCO2',
 'respiratory_rate',
 'oxygen_saturation',
 'FiO2',
 'paO2',
 #'ventilator_mode',
 ]



CLIP_RULES = {
    "bicarbonate": {
        "lower": 5.0, "upper": 50.0, "action": "set_missing",
        "unit": "mEq/L", "note": "Broad ICU-plausible bicarbonate range."
    },
    "PTT": {
        "lower": 18.0, "upper": 150.0, "action": "set_missing",
        "unit": "sec", "note": "Keep prolonged PTT but remove implausible values."
    },
    "temperature": {
        "lower": 30.0, "upper": 43.0, "action": "set_missing",
        "unit": "C", "note": "Outside this is almost certainly artifact/sentinel."
    },
    "hemoglobin": {
        "lower": 2.5, "upper": 25.0, "action": "set_missing",
        "unit": "g/dL", "note": "Absurd or Extreme values likely unit or extraction errors."
    },
    "heart_rate": {
        "lower": 25.0, "upper": 220.0, "action": "set_missing",
        "unit": "bpm", "note": "Keep severe brady/tachycardia, remove impossible monitor artifacts."
    },
    "arterial_blood_pressure": {
        "lower": 20.0, "upper": 180.0, "action": "set_missing",
        "unit": "mmHg", "note": "Broad ICU MAP range."
    },
    "creatinine": {
        "lower": 0.0, "upper": 1500.0, "action": "set_missing",
        "unit": "umol/L", "note": "For µmol/L scale"
    },
    "blood_urea_nitrogen": {
        "lower": 0.3, "upper": 80.0, "action": "set_missing",
        "unit": "mmol/L", "note": "Long right tail can be real in ICU."
    },
    "urine_output": {
        "lower": 0.0, "upper": 2500.0, "action": "set_missing",
        "unit": "aggregated urine metric", "note": "Zero is real, negatives are not."
    },
    "lactate": {
        "lower": 0.3, "upper": 25.0, "action": "set_missing",
        "unit": "mmol/L", "note": "Preserve real ICU hyperlactatemia."
    },
    "paCO2": {
        "lower": 10.0, "upper": 150.0, "action": "set_missing",
        "unit": "mmHg", "note": "0 and extreme values likely artifact."
    },
    "respiratory_rate": {
        "lower": 4.0, "upper": 80.0, "action": "set_missing",
        "unit": "breaths/min", "note": "Retain severe RR, remove impossible values."
    },
    "oxygen_saturation": {
        "lower": 50.0, "upper": 100.0, "action": "set_missing",
        "unit": "%", "note": "Bounded by definition."
    },
    "FiO2": {
        "lower": 21.0, "upper": 100.0, "action": "set_missing",
        "unit": "%", "note": "FiO2 as percent, not fraction."
    },
    "paO2": {
        "lower": 20.0, "upper": 500.0, "action": "set_missing",
        "unit": "mmHg", "note": "Keep hyperoxia but trim implausible far tail."
    },
}
```

```python
def apply_clinical_clipping(df, clip_rules, verbose=True):
    """
    Apply variable-specific clipping / invalid-value-to-missing rules.

    Returns
    -------
    df_out : pd.DataFrame
        Cleaned dataframe
    audit_df : pd.DataFrame
        Summary of what was changed per variable
    """
    df_out = df.copy()
    audit_rows = []

    for var, rule in clip_rules.items():
        if var not in df_out.columns:
            audit_rows.append({
                "variable": var,
                "present": False,
                "action": rule["action"],
                "lower": rule["lower"],
                "upper": rule["upper"],
                "n_nonmissing_before": np.nan,
                "n_below": np.nan,
                "n_above": np.nan,
                "n_changed": np.nan,
                "n_nonmissing_after": np.nan,
            })
            continue

        s = pd.to_numeric(df_out[var], errors="coerce")
        lower, upper = rule["lower"], rule["upper"]
        action = rule["action"]

        below = s < lower
        above = s > upper
        outside = below | above

        n_nonmissing_before = int(s.notna().sum())
        n_below = int(below.sum())
        n_above = int(above.sum())

        s_new = s.copy()

        if action == "set_missing":
            s_new[outside] = np.nan
        elif action == "clip":
            s_new = s_new.clip(lower=lower, upper=upper)
        else:
            raise ValueError(f"Unknown action '{action}' for {var}")

        #n_changed = int((s_new != s).fillna(False).sum() + ((s.isna()) != (s_new.isna())).sum())
        # A value "changed" if either its number changed or its NaN-status flipped.
        changed_mask = (s_new != s).fillna(False) | (s.isna() != s_new.isna())
        n_changed = int(changed_mask.sum())
        
        n_nonmissing_after = int(s_new.notna().sum())

        df_out[var] = s_new

        audit_rows.append({
            "variable": var,
            "present": True,
            "action": action,
            "lower": lower,
            "upper": upper,
            "unit": rule.get("unit"),
            "note": rule.get("note"),
            "n_nonmissing_before": n_nonmissing_before,
            "n_below": n_below,
            "n_above": n_above,
            "n_outside_total": int(outside.sum()),
            "n_changed": n_changed,
            "n_nonmissing_after": n_nonmissing_after,
        })

        if verbose:
            print(
                f"{var}: action={action}, below={n_below}, above={n_above}, "
                f"nonmissing_before={n_nonmissing_before}, nonmissing_after={n_nonmissing_after}"
            )

    audit_df = pd.DataFrame(audit_rows)
    return df_out, audit_df
```

```python
ltm_wide_clipped, clipping_audit = apply_clinical_clipping(
    ltm_wide,
    CLIP_RULES,
    verbose=True
)

display(clipping_audit)
```

```python
def compare_before_after(df_before, df_after, variables, ncols=2, bins=60):
    for var in variables:
        if var not in df_before.columns or var not in df_after.columns:
            continue

        x_before = pd.to_numeric(df_before[var], errors="coerce").dropna()
        x_after = pd.to_numeric(df_after[var], errors="coerce").dropna()

        fig, axes = plt.subplots(1, 2, figsize=(12, 4))

        axes[0].hist(x_before, bins=bins)
        axes[0].set_title(f"{var} - before")

        axes[1].hist(x_after, bins=bins)
        axes[1].set_title(f"{var} - after")

        plt.tight_layout()
        plt.show()

compare_before_after(ltm_wide, ltm_wide_clipped, list(CLIP_RULES.keys()))
```

## Remove stay_ids which do not have data for more than >=12h

```python
# ---------------------------------------------------------------------
# Cohort eligibility: patient must be alive AND in ICU at tau_0 = intime + 12h.
#
# Two predicates:
#   (a) Administratively in ICU at tau_0, with enough room past tau_0 for the
#       leakage buffer:    los >= 12h + LEAK_BUFFER  (i.e., los >= 17h).
#   (b) Alive at tau_0:    death_time_from_intime is NaN OR >= 12h.
#
# Predicate (a) was the only filter in the previous version. Predicate (b)
# was missing, because in MIMIC-IV `outtime` is the bed-clear timestamp,
# not the death time — so `los` can be >= 17h even when the patient died
# in the first 12h. Diagnostic on a prior run found 646 such stays.
# ---------------------------------------------------------------------

MIN_HOURS    = pd.Timedelta(hours=12)
TAU_0_OFFSET = pd.Timedelta(hours=12)
# LEAK_BUFFER is defined earlier in the notebook (cell with drop_near_discharge).

# Per-stay summaries.
los_per_stay   = ltm_wide_confounders.groupby("stay_id")["los"].first()
death_per_stay = ltm_wide_confounders.groupby("stay_id")["death_time_from_intime"].first()
death_per_stay = pd.to_timedelta(death_per_stay, errors="coerce")

# (a) Insufficient ICU bed-occupancy past tau_0 + buffer.
short_los_stays = los_per_stay[(los_per_stay - LEAK_BUFFER) < MIN_HOURS].index

# (b) Death before tau_0.
pre_baseline_death_stays = death_per_stay[
    death_per_stay.notna() & (death_per_stay < TAU_0_OFFSET)
].index

# Union: any stay failing either predicate is excluded.
excluded_stays = short_los_stays.union(pre_baseline_death_stays)

n_short        = len(short_los_stays)
n_pre_baseline = len(pre_baseline_death_stays)
n_both         = len(short_los_stays.intersection(pre_baseline_death_stays))
n_excluded     = len(excluded_stays)

print(f"Stays excluded by short LOS:              {n_short}")
print(f"Stays excluded by pre-baseline death:     {n_pre_baseline}")
print(f"  (of which also short LOS):              {n_both}")
print(f"Total unique stays excluded:              {n_excluded}")
print(f"Stays remaining: "
      f"{ltm_wide_confounders['stay_id'].nunique() - n_excluded}")
```

```python
# Apply the eligibility filter (union of short-LOS and pre-baseline-death exclusions).
ltm_wide_confounders = ltm_wide_confounders[
    ~ltm_wide_confounders["stay_id"].isin(excluded_stays)
].copy()

print("Stays remaining after eligibility filter:",
      ltm_wide_confounders["stay_id"].nunique())
```

# Gridding data for every 12h

```python
# ====================================================================
# 12h gridding — vectorized, schema-consistent
# 
# Schema: W^L_k = [t_k - w_L, t_k), strictly right-open.
#         t_k = k * grid_step_hours hours since intime, for k = 1, 2, ...
#         Grid extent per stay: t_k <= los - leak_buffer.
#
# Pipeline contract (assumed):
#   - Rows within leak_buffer of discharge already dropped (cell 11).
#   - Stays with los < min_usable_hours + leak_buffer already dropped (cell 39).
#   - ventilator_mode NaN already replaced with "unknown" (cell 46).
# ====================================================================

# --- Parameters (single source of truth) ----------------------------
W_L_HOURS         = 12
GRID_STEP_HOURS   = 12
LEAK_BUFFER_H     = 5
MIN_USABLE_HOURS  = 12
UNKNOWN_TOKEN     = "unknown"

VENT_HOURS_COL    = f"vent_mode__hours_since_last__last_{int(W_L_HOURS)}h"

# --- Defensive copies & dtype enforcement ---------------------------
df = ltm_wide_confounders.copy()
df["time"] = pd.to_timedelta(df["time"])
df["los"]  = pd.to_timedelta(df["los"])
df["time_h"] = df["time"].dt.total_seconds() / 3600.0
df["los_h"]  = df["los"].dt.total_seconds()  / 3600.0

# --- Contract assertions -------------------------------------------
_min_los_required = MIN_USABLE_HOURS + LEAK_BUFFER_H
assert (df.groupby("stay_id")["los_h"].first() >= _min_los_required).all(), (
    f"Found stays with los < {_min_los_required}h; "
    f"upstream short-stay filter (cell 39) appears not to have run."
)
assert (df["time_h"] <= df["los_h"] - LEAK_BUFFER_H + 1e-9).all(), (
    f"Found observations within {LEAK_BUFFER_H}h of discharge; "
    f"upstream drop_near_discharge (cell 11) appears not to have run."
)

# --- 1. Assign each observation to its W^L_k window -----------------
# W^L_k = [t_k - w_L, t_k) with t_k = k * step.
# When w_L == step, an obs at time_h belongs to k = floor(time_h / step) + 1,
# which is the right-open convention: obs at exactly time_h = (k-1)*step
# falls in W^L_k (start is included), obs at exactly time_h = k*step falls
# in W^L_{k+1} (end is excluded).
assert W_L_HOURS == GRID_STEP_HOURS, (
    "Generalised path for non-equal w_L and grid_step is not implemented."
)
df["k"] = np.floor(df["time_h"] / GRID_STEP_HOURS).astype(int) + 1

# --- 2. Per-stay valid grid extent ----------------------------------
# t_k <= los - leak_buffer  =>  k <= floor((los - buffer) / step)
los_per_stay   = df.groupby("stay_id")["los_h"].first()
k_max_per_stay = np.floor(
    (los_per_stay - LEAK_BUFFER_H) / GRID_STEP_HOURS
).astype(int)
k_max_per_stay = k_max_per_stay[k_max_per_stay >= 1]   # require at least t_1

if k_max_per_stay.empty:
    raise RuntimeError("No stays produced any valid grid points; check filters.")

# Drop obs whose k exceeds the stay's k_max (these are post-valid-grid).
df = df.merge(
    k_max_per_stay.rename("k_max").reset_index(), on="stay_id", how="inner"
)
df = df[df["k"] <= df["k_max"]]

# --- 3. Aggregate covariates per (stay_id, k) -----------------------
g = df.groupby(["stay_id", "k"], sort=True)

# Means (numeric, skipna by default)
means = g[mean_vars].mean()

# Observation count per window — diagnostic
n_obs = g.size().rename("n_obs_in_window")

# Last non-`unknown` per categorical/numeric `last_vars`
#   For numeric columns, `!= "unknown"` is always True (no-op), so the same
#   logic works uniformly.
last_results = {}
all_keys = g.size().index   # full set of (stay_id, k) keys observed in data
for vname in last_vars:
    mask = df[vname].ne(UNKNOWN_TOKEN) & df[vname].notna()
    sub = df.loc[mask, ["stay_id", "k", "time_h", vname]]
    if sub.empty:
        last_results[vname] = pd.Series(index=all_keys, dtype="object", name=vname)
    else:
        idx = sub.groupby(["stay_id", "k"])["time_h"].idxmax()
        last_results[vname] = (
            sub.loc[idx]
               .set_index(["stay_id", "k"])[vname]
               .reindex(all_keys)
        )
last_df = pd.DataFrame(last_results)

# vent_mode "hours since last known recording" within W^L_k
vent_mask = (
    df[VENT_HOURS_COL.split("__")[0].replace("vent_mode", "ventilator_mode")]  # = ventilator_mode
    .ne(UNKNOWN_TOKEN) & df["ventilator_mode"].notna()
)
# (simpler equivalent of the line above:)
vent_mask = df["ventilator_mode"].ne(UNKNOWN_TOKEN) & df["ventilator_mode"].notna()
vent_sub = df.loc[vent_mask, ["stay_id", "k", "time_h"]]
if vent_sub.empty:
    hours_since = pd.Series(np.nan, index=all_keys, name=VENT_HOURS_COL)
else:
    last_time_per_kg = vent_sub.groupby(["stay_id", "k"])["time_h"].max()
    k_index = last_time_per_kg.index.get_level_values("k").astype(float)
    t_k_h = k_index * GRID_STEP_HOURS
    diff = np.clip(t_k_h - last_time_per_kg.values, 0, W_L_HOURS)
    hours_since = (
        pd.Series(diff, index=last_time_per_kg.index, name=VENT_HOURS_COL)
          .reindex(all_keys)
    )

# --- 4. Build the full (stay_id, k) grid, including empty windows ---
# A stay should emit one row per k in [1, k_max], even if W^L_k contains
# no observations (then all aggregates are NaN).
full_grid = (
    k_max_per_stay.rename("k_max").reset_index()
    .assign(k=lambda x: x["k_max"].apply(lambda km: np.arange(1, km + 1)))
    .explode("k")
    .astype({"k": int})
    .drop(columns="k_max")
    .set_index(["stay_id", "k"])
)

# --- 5. Assemble aggregates onto the full grid ----------------------
out = (
    full_grid
      .join(means, how="left")
      .join(last_df, how="left")
      .join(hours_since, how="left")
      .join(n_obs, how="left")
      .reset_index()
)

# --- 6. Merge baseline (one row per stay, time-invariant) -----------
baseline_df = (
    df.sort_values(["stay_id", "time_h"])
      .drop_duplicates("stay_id", keep="first")
      [baseline]                                    # `baseline` includes stay_id
      .reset_index(drop=True)
)
# Drop stay_id from `baseline_df` columns we already have, to avoid suffix mess
baseline_extras = [c for c in baseline_df.columns if c != "stay_id"]
out = out.merge(baseline_df[["stay_id"] + baseline_extras], on="stay_id", how="left")

# --- 7. Add `grid_end` as Timedelta (downstream cells expect this) --
out["grid_end"] = pd.to_timedelta(out["k"] * GRID_STEP_HOURS, unit="h")

# --- 8. Reorder columns to match the original notebook output -------
front = ["stay_id", "k", "grid_end"]
rest  = [c for c in out.columns if c not in front]
ltm_grided = out[front + rest].sort_values(["stay_id", "k"]).reset_index(drop=True)

# --- 9. Post-hoc sanity assertions ----------------------------------
_los_h = ltm_grided["stay_id"].map(los_per_stay)
_grid_end_h = ltm_grided["grid_end"].dt.total_seconds() / 3600.0
assert (_grid_end_h <= _los_h - LEAK_BUFFER_H + 1e-9).all(), (
    "Some grid_end values violate the leakage-buffer constraint."
)
assert (
    ltm_grided.groupby("stay_id")["k"]
              .apply(lambda s: (s.values == np.arange(1, s.max() + 1)).all())
              .all()
), "Non-contiguous k within some stays."

print(f"Gridded {ltm_grided['stay_id'].nunique():,} stays into "
      f"{len(ltm_grided):,} (stay, k) rows.")
```
