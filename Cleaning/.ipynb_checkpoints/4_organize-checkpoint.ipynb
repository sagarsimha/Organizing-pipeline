## Death times

```python
# ---------------------------------------------------------------------
# Compute absolute death time per row.
#
# Convention notes:
#   - tau_0 = intime + 12h is the analysis baseline (see 3_organize).
#     The in-ICU mortality interval is [tau_0, outtime].
#   - The right boundary at outtime is RIGHT-CLOSED. This is deliberate
#     and *intentionally* differs from the right-open conventions used
#     for W^L_t and W^Y_t. In MIMIC-IV, for in-ICU deaths, outtime is
#     typically equal to the death time, so death_abs_time == outtime is
#     the modal case for true in-ICU deaths. Switching to a right-open
#     boundary here would exclude essentially every in-ICU death.
# ---------------------------------------------------------------------

# Ensure correct dtypes.
ltm_clipped_grided["intime"]  = pd.to_datetime(ltm_clipped_grided["intime"],  errors="coerce")
ltm_clipped_grided["outtime"] = pd.to_datetime(ltm_clipped_grided["outtime"], errors="coerce")
ltm_clipped_grided["death_time_from_intime"] = pd.to_timedelta(
    ltm_clipped_grided["death_time_from_intime"], errors="coerce"
)

# Absolute death time (NaT where no death recorded).
ltm_clipped_grided["death_abs_time"] = (
    ltm_clipped_grided["intime"] + ltm_clipped_grided["death_time_from_intime"]
)
```

## Derive icu_mortality from timestamp

```python
# ---------------------------------------------------------------------
# Derive icu_mortality from timestamps.
#
# An in-ICU death is one where death_abs_time falls in [tau_0, outtime].
# The lower bound is tau_0 = intime + 12h, NOT intime, to align with the
# analysis baseline (see 3_organize).
#
# Pre-baseline deaths (in [intime, tau_0)) should not exist after the
# upstream LOS filter (3_organize cell 39: stays with los < 17h dropped),
# but we assert this rather than rely on it implicitly.
# ---------------------------------------------------------------------

TAU_0_OFFSET = pd.Timedelta(hours=12)
_tau_0 = ltm_clipped_grided["intime"] + TAU_0_OFFSET

ltm_clipped_grided["icu_mortality_derived"] = (
    ltm_clipped_grided["death_abs_time"].notna()
    & (ltm_clipped_grided["death_abs_time"] >= _tau_0)
    & (ltm_clipped_grided["death_abs_time"] <= ltm_clipped_grided["outtime"])
)

# Assert no pre-baseline deaths slipped through the upstream LOS filter.
_pre_baseline = (
    ltm_clipped_grided["death_abs_time"].notna()
    & (ltm_clipped_grided["death_abs_time"] >= ltm_clipped_grided["intime"])
    & (ltm_clipped_grided["death_abs_time"] < _tau_0)
)
_n_pre_baseline_stays = (
    ltm_clipped_grided.loc[_pre_baseline, "stay_id"].nunique()
)
assert _n_pre_baseline_stays == 0, (
    f"{_n_pre_baseline_stays} stays have a death in [intime, tau_0). "
    f"Upstream LOS filter (3_organize cell 39) appears not to have run, "
    f"or the LOS threshold has been relaxed below 12h + leak_buffer."
)
```

```python
ltm_clipped_grided#ratio of mortality
last_rows_der = ltm_clipped_grided.groupby("stay_id").tail(1)
(last_rows_der["icu_mortality_derived"] == 1).mean()
```

```python
# ---------------------------------------------------------------------
# Compare the precomputed `icu_mortality` label against the
# timestamp-derived flag. Any disagreement is a data-quality signal that
# deserves inspection before deciding which column to treat as ground
# truth for D.
#
# Note: 5_organize renames `icu_mortality_derived` -> `icu_mortality`
# (and the original label -> `icu_mortality_flag`), so the
# timestamp-derived version is the source of truth downstream. The
# diagnostic below makes that choice auditable.
# ---------------------------------------------------------------------

# Per-stay collapse: a stay is "label-positive" if the precomputed
# icu_mortality is True for any of its rows; similarly for derived.
per_stay = (
    ltm_clipped_grided.groupby("stay_id")
    .agg(label=("icu_mortality",         "max"),
         derived=("icu_mortality_derived", "max"))
)
per_stay["label"]   = per_stay["label"].astype(bool)
per_stay["derived"] = per_stay["derived"].astype(bool)

# Confusion table.
xt = pd.crosstab(
    per_stay["label"], per_stay["derived"],
    rownames=["icu_mortality (label)"],
    colnames=["icu_mortality_derived"],
    dropna=False,
)
print("Stay-level agreement table:")
print(xt)

# Targeted breakdown of disagreements.
disagree_stays = per_stay.index[per_stay["label"] != per_stay["derived"]]
n_disagree = len(disagree_stays)
print(f"\n{n_disagree} stays disagree between label and derived.")

if n_disagree > 0:
    diag = (
        ltm_clipped_grided.loc[
            ltm_clipped_grided["stay_id"].isin(disagree_stays),
            ["stay_id", "icu_mortality", "icu_mortality_derived",
             "intime", "outtime", "death_abs_time", "death_time_from_intime"]
        ]
        .drop_duplicates("stay_id")
    )

    # Categorise: which side is positive?
    diag["case"] = np.select(
        [
            diag["icu_mortality"].astype(bool)        & ~diag["icu_mortality_derived"],
            ~diag["icu_mortality"].astype(bool)       &  diag["icu_mortality_derived"],
        ],
        [
            "label_only (label says died, timestamp does not place in [tau_0, outtime])",
            "derived_only (timestamp places death in [tau_0, outtime], label says alive)",
        ],
        default="other",
    )
    print("\nDisagreement breakdown:")
    print(diag["case"].value_counts())
    print("\nExamples:")
    print(diag.head(10).to_string(index=False))
```

# Converting fio2 and GCS into ordinal bins

```python
# Modify fio2__last__last_12h
def transform_fio2_to_ordered_bins(df, col="fio2__last__last_12h"):
    df = df.copy()

    # Only valid after clipping to [21, 100]
    bins = [20.999, 30, 40, 50, 60, 80, 100]
    labels = [
        "21_30",
        "31_40",
        "41_50",
        "51_60",
        "61_80",
        "81_100"
    ]

    df[col] = pd.cut(
        df[col],
        bins=bins,
        labels=labels,
        include_lowest=True,
        ordered=True
    )

    # Add missing categories
    #df[col] = df[col].cat.add_categories(["missing"])
    #df[col] = df[col].fillna("missing")

    return df

ltm_clipped_grided = transform_fio2_to_ordered_bins(ltm_clipped_grided)
```

```python
print(ltm_clipped_grided["fio2__last__last_12h"].dtype)
print(ltm_clipped_grided["fio2__last__last_12h"].cat.categories)
print(ltm_clipped_grided["fio2__last__last_12h"].value_counts(dropna=False))
```

```python
def transform_gcs_to_4bins(df, col="glasgow_coma_scale_total__last__last_12h"):
    df = df.copy()

    # Define bins
    bins = [-np.inf, 8, 12, 14, 15]
    labels = [
        "severe",          # 3–8
        "moderate",        # 9–12
        "mild_impaired",   # 13–14
        "normal"           # 15
    ]

    # Apply ordered categorical binning
    df[col] = pd.cut(
        df[col],
        bins=bins,
        labels=labels,
        ordered=True,
        right=True  # include upper bound
    )

    return df
ltm_clipped_grided = transform_gcs_to_4bins(ltm_clipped_grided)
```

```python
print(ltm_clipped_grided["glasgow_coma_scale_total__last__last_12h"].dtype)
print(ltm_clipped_grided["glasgow_coma_scale_total__last__last_12h"].cat.categories)
print(ltm_clipped_grided["glasgow_coma_scale_total__last__last_12h"].value_counts(dropna=False))
```

# Plotting distributions

```python
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# Define variables
# -----------------------------
cat_vars = [
    "vent_mode__last__last_12h",
    "fio2__last__last_12h",
    "glasgow_coma_scale_total__last__last_12h"
]

num_vars = [v for v in L_confounders if v not in cat_vars]

# -----------------------------
# NUMERIC VARIABLES: KDE plots
# -----------------------------
n = len(num_vars)
cols = 4
rows = (n + cols - 1) // cols

plt.figure(figsize=(20, rows * 3))

for i, var in enumerate(num_vars, 1):
    plt.subplot(rows, cols, i)
    
    sns.kdeplot(
        data=ltm_clipped_grided,
        x=var,
        fill=True,
        bw_adjust=1
    )
    
    plt.title(var)

plt.tight_layout()
plt.savefig("./numericals.pdf")
plt.show()


# -----------------------------
# CATEGORICAL VARIABLES: Countplots
# -----------------------------
n_cat = len(cat_vars)
cols = 2
rows = (n_cat + cols - 1) // cols

plt.figure(figsize=(12, rows * 4))

for i, var in enumerate(cat_vars, 1):
    plt.subplot(rows, cols, i)
    
    # Ensure consistent ordering for ordinal variables
    order = sorted(ltm_clipped_grided[var].dropna().unique())
    
    sns.countplot(
        data=ltm_clipped_grided,
        x=var,
        order=order
    )
    
    plt.xticks(rotation=45)
    plt.title(var)

plt.tight_layout()
plt.savefig("./categoricals.pdf")
plt.show()
```

```python
baseline = ["raw_age", 
            "sex", 
            "raw_height", 
            "raw_weight", 
            "unit_type", 
            "origin",
            "subject_id",
            "intime","outtime","los","death_time_from_intime","icu_mortality_derived"]

def fill_baseline_first_valid(df, baseline_vars):
    """Fill baseline vars per stay with first non-NaN value."""
    for col in baseline_vars:
        df[col] = df.groupby("stay_id")[col].transform("first")
    return df
ltm_clipped_grided = fill_baseline_first_valid(ltm_clipped_grided, baseline)
```
