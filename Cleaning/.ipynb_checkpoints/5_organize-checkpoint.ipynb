1. Remove admission_ids which have both icu_mortality and mortality_after_discharge as True
3. [Done before gridding which is more appropriate] Remove row in each admission which are very close (<=5 hrs) to discharge. This is to avoid data leakage on discharge decision. Check if this can be done at 3_organizing pipeline.
4. Add A, D
5. Add Y (All-cause mortality - in-icu mortality or mortality after discharge)

# Cleaning (corrected)

This notebook is a corrected version of the previous `5_organize.ipynb`. The schema is the one fixed in `3_organize`: covariate windows $W^L_t = [t_k - 12, t_k)$, outcome windows $W^Y_t = [t_k, t_{k+1})$, decision grid $t_k = (t+1) \cdot 12\text{h}$ since intime, with $\tau_0 = \text{intime} + 12\text{h}$ and end-of-follow-up $K = \tau_0 + 30\text{d}$.

**Fixes applied here:**

1. **`t_cutoff` corrected from 60 to 59.** The previous version kept a final decision row at $t = 60$ whose outcome window $W^Y_{60} = [732, 744)\text{h}$ ran past $K = 732\text{h}$. The last valid decision is $t = 59$ with $W^Y_{59} = [720, 732)\text{h}$, exactly closing at $K$.
4. **`_t_cutoff` derived once per stay** rather than via a per-row groupby-lambda. Faster and not silently fragile to future per-row intime drift.
5. **Admin-censoring is now identified explicitly by a `_admin_censored` flag** emitted by `truncate_followup_30d_t`, instead of being inferred downstream from `t == T_MAX`. The previous heuristic mis-classified any natural discharge or in-ICU death occurring exactly at $t = 59$.
6. **Terminal-row encoding consolidated.** The previous overlap between cell 30 (pre-truncation partial assignment) and cell 33 (post-truncation reassignment) is removed. Cell 30 now only initialises defaults; truncation only filters and flags; the post-truncation block is the single source of truth for terminal-row $A, D, Z, Y$. The result is idempotent under repeated execution.
7. **Invariant assertions added** (new cell) that pin the encoding contract: $D = \text{NaN}$ iff $A = 1$ at terminal row, $Z$ non-missing iff $A = 1$, non-terminal rows are zeros/NaNs, $t \leq T_\text{MAX} = 59$, and the observed last `grid_end` does not exceed $720\text{h}$ from intime.

**Retained by user decision (not changed):**

2. The post-discharge mortality window remains anchored to admission ($Z = \mathbb{1}\{\text{outtime} < \text{death} \leq \tau_0 + 30\text{d}\}$) rather than to discharge.  This induces informative window-length variation across discharge times; this is a known limitation to be disclosed/explored in sensitivity analysis, not a coding issue.
3. The $D = \text{NaN}$ encoding on discharged terminal rows is **correct**, representing "in-ICU death not at risk because $A=1$ absorbed the patient out of the ICU." Downstream $D$-model fitting must drop NaN rows; the same row still contributes to the $A$- and $Z$-models.

## Mortality within X days after discharge

```python
# Ensure correct dtypes
sf_mimicdata["intime"]  = pd.to_datetime(sf_mimicdata["intime"], errors="coerce")
sf_mimicdata["outtime"] = pd.to_datetime(sf_mimicdata["outtime"], errors="coerce")

# death_time_from_intime must be timedelta
sf_mimicdata["death_time_from_intime"] = pd.to_timedelta(
    sf_mimicdata["death_time_from_intime"], errors="coerce"
)

# Absolute death time = intime + death offset
sf_mimicdata["death_abs_time"] = sf_mimicdata["intime"] + sf_mimicdata["death_time_from_intime"]

# Post-discharge mortality [outtime < death_abs_time ≤ (intime+12h)+30d]
HORIZON = pd.Timedelta(days=30)
sf_mimicdata["mortality_after_discharge_30d"] = (
    sf_mimicdata["death_abs_time"].notna()
    & (sf_mimicdata["death_abs_time"] > sf_mimicdata["outtime"])
    & (sf_mimicdata["death_abs_time"] <=
       (sf_mimicdata["intime"] + pd.Timedelta(hours=12) + HORIZON))
).astype(int)
```

```python
#ratio of mortality
last_rows = sf_mimicdata.groupby("stay_id").tail(1)
(last_rows["mortality_after_discharge_30d"] == 1).mean()
```

# Add A, Y and D

```python
# Initialise outcome columns on ALL rows.
# Terminal-row encoding is performed *after* truncation, in a single
# deterministic block below (issue 6). Non-terminal rows retain these
# defaults end-to-end.
sf_mimicdata.loc[:, "A"] = 0
sf_mimicdata.loc[:, "D"] = 0
sf_mimicdata.loc[:, "Z"] = np.nan
sf_mimicdata.loc[:, "Y"] = np.nan

# Ensure stable ordering for subsequent terminal-row operations.
sf_mimicdata = sf_mimicdata.sort_values(["stay_id", "t"]).copy()
```

# Limit the study to 30 days since admission. 
# For stay_ids which have LOS lesser than 30days since admission, keep them as is.
# For stays which are more than 30 days since admission, chop the rows to have LOS at 30 days only and declare those stays to have D=0, Z=0, Y=0 at the last row

```python
def truncate_followup_30d_t(
    df,
    id_col="stay_id",
    t_col="t",
    intime_col="intime",
    bin_hours=12,
    horizon_days=30,
    baseline_offset_hours=12,
):
    """
    Truncate longitudinal ICU data at follow-up end K = intime + 12h + 30d.

    Schema recap (matches 3_organize):
      tau_0 = intime + 12h        (anchor time, first decision)
      t = 0 corresponds to grid_end = 12h since intime, i.e. tau_0.
      W^Y_t  = [t_k, t_{k+1})     is the outcome accrual window for decision t.

    The last *decision* point t_max is the largest t for which W^Y_t lies
    inside [tau_0, K]:
        grid_end_h(t) = (t + 1) * bin_hours
        require grid_end_h(t) + bin_hours <= K_h
        => t <= (K_h - bin_hours) / bin_hours - 1
        with K_h = baseline_offset_hours + horizon_days * 24
        and bin_hours = baseline_offset_hours = 12  => t_max = 59.

    The previous version used `t_cutoff = 60` (i.e. floor(K/12) - 1 with
    K_h = 732), which kept one decision point whose outcome window W^Y_t
    would have run past K. That row has been removed.

    Design note (issue 6):
      This function performs ONLY truncation. It does NOT assign A/D/Z/Y
      on the truncation boundary. It returns a boolean column
      `_admin_censored` on the kept rows that flags the row as the
      truncation boundary (True) versus a natural terminal row (False).
      Downstream code (cell 34) is the single source of truth for
      terminal-row outcome encoding. This removes the previous overlap
      between this function's assignments and the cell-33 reassignments.

    Parameters
    ----------
    df : DataFrame
        Long-format data with one row per (stay_id, t).
    id_col, t_col, intime_col : str
    bin_hours : int
        Width of each discrete time interval (default 12h).
    horizon_days : int
        Follow-up horizon past tau_0 (default 30 days).
    baseline_offset_hours : int
        Hours from intime to tau_0 (default 12h).

    Returns
    -------
    DataFrame, truncated, with an added boolean column `_admin_censored`.
    """
    df = df.copy()
    df[intime_col] = pd.to_datetime(df[intime_col], errors="coerce")
    df = df.sort_values([id_col, t_col]).reset_index(drop=True)

    # ----- derive t_max (constant across stays under fixed schema) --------
    K_h = baseline_offset_hours + horizon_days * 24                # = 732
    # Closed form: t_max = (K_h - bin_hours) / bin_hours - 1.
    # With defaults: (732 - 12)/12 - 1 = 59.
    t_max_global = (K_h - bin_hours) // bin_hours - 1              # = 59

    # Per-stay t_cutoff is identical when intime is constant per stay
    # (true by construction). Compute once per stay (issue 4).
    stay_ids = df[id_col].unique()
    t_cutoff_per_stay = pd.Series(t_max_global, index=stay_ids, name="_t_cutoff")
    df["_t_cutoff"] = df[id_col].map(t_cutoff_per_stay)

    # Identify long stays.
    max_t_per_stay = df.groupby(id_col)[t_col].transform("max")
    df["_long"] = max_t_per_stay > df["_t_cutoff"]

    # Keep rows within cutoff.
    kept = df[(~df["_long"]) | (df[t_col] <= df["_t_cutoff"])].copy()

    # Admin-censoring flag: True for the truncation boundary row of long
    # stays, False everywhere else (including natural discharge or death
    # exactly at t = t_max in a short/medium stay).
    kept["_admin_censored"] = kept["_long"] & (kept[t_col] == kept["_t_cutoff"])

    kept = kept.drop(columns=["_t_cutoff", "_long"], errors="ignore")
    kept = kept.sort_values([id_col, t_col]).reset_index(drop=True)

    return kept
```

```python
sf_mimicdata_30d = truncate_followup_30d_t(sf_mimicdata)

# =====================================================================
# Final terminal-row encoding (single source of truth — issue 6).
#
# Encoding convention (issue 3 retained):
#   Non-terminal rows:       A=0, D=0, Z=NaN, Y=NaN
#   Terminal — in-ICU death: A=0, D=1, Z=NaN,        Y=1
#   Terminal — discharged:   A=1, D=NaN, Z in {0,1}, Y=Z
#   Terminal — admin-cens.:  A=0, D=0,  Z=NaN,       Y=0
#
# D=NaN on discharged rows is the correct encoding of "in-ICU death not
# at risk because A=1 absorbed the patient out of the ICU." Downstream
# nuisance models for D must drop NaN rows (standard pandas/LGBM
# behaviour); the row still contributes to the A and Z models.
# =====================================================================

# Sort once for stable terminal-row identification.
sf_mimicdata_30d = sf_mimicdata_30d.sort_values(["stay_id", "t"]).reset_index(drop=True)

# Index of the terminal row per stay (after truncation).
last_idx_30d = sf_mimicdata_30d.groupby("stay_id", sort=False).tail(1).index

# Pull the needed terminal-row columns for vectorised assignment.
# `_admin_censored` was emitted by truncate_followup_30d_t (cell 32).
term = sf_mimicdata_30d.loc[last_idx_30d, [
    "stay_id", "t", "icu_mortality", "mortality_after_discharge_30d", "_admin_censored"
]].copy()

# Three mutually exclusive terminal-row types, derived from the truncation
# flag and the precomputed icu_mortality:
#   admin-censored  : truncation boundary of a long stay
#   in-ICU death    : icu_mortality is True (and not admin-censored)
#   discharge       : otherwise (patient left ICU alive within follow-up)
# Note we use _admin_censored as the discriminator, NOT `t == T_MAX`,
# because a patient can naturally discharge or die at t = T_MAX in a
# stay that did not require truncation.
term["_is_admin_cens"] = term["_admin_censored"].astype(bool)
term["_is_icu_death"]  = (~term["_is_admin_cens"]) & term["icu_mortality"].astype(bool)
term["_is_discharge"]  = (~term["_is_admin_cens"]) & (~term["_is_icu_death"])

# Sanity: every terminal row is exactly one of the three types.
_one_hot = term[["_is_icu_death", "_is_admin_cens", "_is_discharge"]].sum(axis=1)
assert (_one_hot == 1).all(), "Terminal row classification is not mutually exclusive."

# ----- Assign A, D, Z, Y on terminal rows -----------------------------
# Non-terminal rows keep the defaults A=0, D=0, Z=NaN, Y=NaN from cell 30.
# We first reset terminal rows to defaults, then overwrite per category.
# This is idempotent under repeated execution of this cell.
sf_mimicdata_30d.loc[last_idx_30d, "A"] = 0
sf_mimicdata_30d.loc[last_idx_30d, "D"] = 0
sf_mimicdata_30d.loc[last_idx_30d, "Z"] = np.nan
sf_mimicdata_30d.loc[last_idx_30d, "Y"] = np.nan

# (i) In-ICU death.
mask_icu_death = term["_is_icu_death"].values
sf_mimicdata_30d.loc[last_idx_30d[mask_icu_death], "A"] = 0
sf_mimicdata_30d.loc[last_idx_30d[mask_icu_death], "D"] = 1
sf_mimicdata_30d.loc[last_idx_30d[mask_icu_death], "Z"] = np.nan
sf_mimicdata_30d.loc[last_idx_30d[mask_icu_death], "Y"] = 1

# (ii) Discharge.
mask_disch = term["_is_discharge"].values
sf_mimicdata_30d.loc[last_idx_30d[mask_disch], "A"] = 1
sf_mimicdata_30d.loc[last_idx_30d[mask_disch], "D"] = np.nan
sf_mimicdata_30d.loc[last_idx_30d[mask_disch], "Z"] = (
    term.loc[mask_disch, "mortality_after_discharge_30d"].values
)
sf_mimicdata_30d.loc[last_idx_30d[mask_disch], "Y"] = (
    term.loc[mask_disch, "mortality_after_discharge_30d"].values
)

# (iii) Administrative censoring (long stay truncated at t = T_MAX).
mask_admin = term["_is_admin_cens"].values
sf_mimicdata_30d.loc[last_idx_30d[mask_admin], "A"] = 0
sf_mimicdata_30d.loc[last_idx_30d[mask_admin], "D"] = 0
sf_mimicdata_30d.loc[last_idx_30d[mask_admin], "Z"] = np.nan
sf_mimicdata_30d.loc[last_idx_30d[mask_admin], "Y"] = 0

# Drop the bookkeeping flag — it has served its purpose.
sf_mimicdata_30d = sf_mimicdata_30d.drop(columns=["_admin_censored"])

T_MAX = sf_mimicdata_30d["t"].max()   # = 59 under default schema
print(f"T_MAX = {T_MAX} (expected 59 under default schema).")
print(f"Terminal-row breakdown: "
      f"in-ICU death = {int(term['_is_icu_death'].sum())}, "
      f"discharge = {int(term['_is_discharge'].sum())}, "
      f"admin-censored = {int(term['_is_admin_cens'].sum())}.")
```

```python
# =====================================================================
# Invariant assertions (cheap, run once).
#
# These pin down the encoding convention agreed for issue 3 and surface
# any future drift (e.g. a future re-ordering of cells silently breaking
# the contract).
# =====================================================================

# Terminal rows only.
_term = sf_mimicdata_30d.groupby("stay_id").tail(1)

# Invariant 1: D is NaN iff A == 1 at terminal row.
assert (_term["D"].isna() == _term["A"].eq(1)).all(), (
    "D=NaN should occur exactly when A=1 at terminal row "
    "(D is undefined post-discharge by construction)."
)

# Invariant 2: Z is non-NaN iff A == 1 at terminal row.
assert (_term["Z"].notna() == _term["A"].eq(1)).all(), (
    "Z should be defined exactly when A=1 at terminal row."
)

# Invariant 3: At most one absorbing event per stay — A=1 XOR D=1 (or neither, if admin-censored).
_absorbed = _term["A"].eq(1).astype(int) + _term["D"].eq(1).fillna(False).astype(int)
assert _absorbed.le(1).all(), (
    "Patient cannot be both discharged (A=1) and in-ICU dead (D=1) at terminal row."
)

# Invariant 4: Non-terminal rows have A=0, D=0, Z=NaN, Y=NaN.
_non_term = (
    sf_mimicdata_30d
    .groupby("stay_id", group_keys=False)
    .apply(lambda g: g.iloc[:-1])
)
assert _non_term["A"].eq(0).all(),  "Non-terminal A must be 0."
assert _non_term["D"].eq(0).all(),  "Non-terminal D must be 0."
assert _non_term["Z"].isna().all(), "Non-terminal Z must be NaN."
assert _non_term["Y"].isna().all(), "Non-terminal Y must be NaN."

# Invariant 5: Y matches the absorbing-event convention. Cross-tab for inspection:
#   in-ICU death  -> A=0, D=1, Z=NaN, Y=1
#   discharge     -> A=1, D=NaN, Z in {0,1}, Y=Z
#   admin-cens.   -> A=0, D=0, Z=NaN, Y=0
_xt = (
    _term.groupby(["A", "D", "Z", "Y"], dropna=False).size().reset_index(name="n")
)
print("Terminal-row (A,D,Z,Y) cross-tab:")
print(_xt)

# Invariant 6: No t exceeds T_MAX = 59 under default schema.
assert sf_mimicdata_30d["t"].max() <= T_MAX, (
    f"Found t > {T_MAX}; truncation did not enforce K-bound."
)

# Invariant 7: Decision-horizon consistency.
# Last grid_end = (T_MAX + 1) * 12h = 720h = 30d from intime.
# Last outcome window W^Y_{T_MAX} = [720h, 732h) reaches exactly K.
_expected_last_grid_end_h = (T_MAX + 1) * 12
_actual_last_grid_end_h = (
    sf_mimicdata_30d["grid_end"].dt.total_seconds().max() / 3600.0
)
assert _actual_last_grid_end_h <= _expected_last_grid_end_h + 1e-9, (
    f"Observed last grid_end ({_actual_last_grid_end_h}h) exceeds expected "
    f"({_expected_last_grid_end_h}h) under K = intime + 12h + 30d."
)
print(f"Schema OK: last grid_end = {_actual_last_grid_end_h}h, T_MAX = {T_MAX}.")
```
